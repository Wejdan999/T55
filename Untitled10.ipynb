```python
from geopy.geocoders import Nominatim

def get_wind_force_category(windspeed: float) -> str:
    if windspeed < 1:
        return "Calm"
    elif 1 <= windspeed <= 5:
        return "Light Air"
    elif 6 <= windspeed <= 11:
        return "Light Breeze"
    elif 12 <= windspeed <= 19:
        return "Gentle Breeze"
    elif 20 <= windspeed <= 28:
        return "Moderate Breeze"
    elif 29 <= windspeed <= 38:
        return "Fresh Breeze"
    elif 39 <= windspeed <= 49:
        return "Strong Breeze"
    elif 50 <= windspeed <= 61:
        return "Gale"
    elif 62 <= windspeed <= 74:
        return "Strong Gale"
    else:
        return "Storm"

def get_weather_by_city(city_name: str):
    # Initialize geocoder
    geolocator = Nominatim(user_agent="LLMexercise")

    # Get location data (latitude and longitude) for the city
    location = geolocator.geocode(city_name)

    if location:
        latitude = location.latitude
        longitude = location.longitude
    else:
        return f"City '{city_name}' not found."

    # Open-Meteo API endpoint with the obtained latitude and longitude
    url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current_weather=true"

    # Send a GET request to the Open-Meteo API
    response = requests.get(url)

    # Check if the request was successful
    if response.status_code == 200:
        # Parse the JSON data from the response
        weather_data = response.json()

        # Extract relevant information from the response
        current_weather = weather_data.get('current_weather', {})
        temperature = current_weather.get('temperature', 'N/A')
        windspeed = current_weather.get('windspeed', 'N/A')
        winddirection = current_weather.get('winddirection', 'N/A')
        weather_time = current_weather.get('time', 'N/A')

        # Check wind force category
        wind_force = get_wind_force_category(windspeed)

        # Return the weather information as a formatted string
        return (
            f"Current weather in {city_name}:\n"
            f"Temperature: {temperature}°C\n"
            f"Wind Speed: {windspeed} km/h\n"
            f"Wind Direction: {winddirection}°\n"
            f"Wind Force: {wind_force}\n"
            f"Time of data: {weather_time}"
        )
    else:
        return "Failed to retrieve weather data."

# Example usage
city_name = "Riyadh, Saudi Arabia"
weather_info = get_weather_by_city(city_name)
print(weather_info)
```

```python

from langchain.agents import Tool

wind_force = Tool(
    name="wind_force",
    func=get_wind_force_category,
    description="Gets the wind force speed."
)
```

```python
from langchain.agents import initialize_agent, AgentType

# List of tools the agent can use
tools = [wind_force]

# Initialize the agent
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent_type=AgentType.ZERO_SHOT_REACT_DESCRIPTION
)
```

```python
def handle_query(query: str) -> str:

    city_name = query.replace("What is the speed force in ", "").replace("?", "").strip()
    return get_weather_by_city(city_name)
```

```python
query = "What is the speed force in Riyadh Saudi Arabia?"
response = handle_query(query)
print(response)
```
